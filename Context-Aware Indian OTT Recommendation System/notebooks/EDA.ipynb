```python
df.replace("-",pd.NA,inplace=True)
```

```python
df.isnull().sum()
```

```python
df.drop("ID",axis=1,inplace=True)
```

```python
df["Year"]=pd.to_numeric(df["Year"],errors="coerce")
df=df[df["Year"].notnull()]
```

```python
df["Rating(10)"]=pd.to_numeric(df["Rating(10)"],errors="coerce")
df["Rating(10)"]=df["Rating(10)"].fillna(df["Rating(10)"].median())
```

```python
df["Votes"]=df["Votes"].fillna(0)
df["Genre"]=df["Genre"].fillna("unknown")
```

```python
df.isnull().sum()
```

```python
df["Votes"].head()
```

```python
df["Votes"]=df["Votes"].astype(str)
```

```python
df["Votes"]=df["Votes"].str.replace(",","")
```

```python
df["Votes"]=pd.to_numeric(df["Votes"],errors="coerce")
```

```python
df["Votes"] = df["Votes"].fillna(0)
df["Votes"]=df["Votes"].astype(int)
```

```python
df["Year"]=df["Year"].astype(int)
```

```python
df["Genre"]=df["Genre"].astype(str)
df["Genre"]=(df["Genre"].str.lower().str.strip())
```

```python
df.info()
```

```python
df["Language"]=df["Language"].replace("oriya","odia")
df["Language"]=df["Language"].replace("rajastani","rajasthani")
```

```python
df["Language"].value_counts()
```

```python
all_genres = (
    df["Genre"]
    .str.split(",")
    .explode()
    .str.strip()
)

print(all_genres.value_counts())
```

```python
df.to_csv("cleaned movies.csv",index=False)
```
